```python
#1. Load the file. Show the shape, the first 5 rows, and info().
df = pd.read_csv('Foodpanda Analysis Dataset (1) - Foodpanda Analysis Dataset (1).csv')
df.head()
```

```python
df["price"] = df["price"].fillna(df["price"].median())
df["rating"] = df["rating"].fillna(df["rating"].mean())
df["category"] = df["category"].fillna(df["category"].mode()[0])
```

```python
df["total_sales"] = df["price"] * df["quantity"]
df.head()
```

```python
df.groupby("dish_name")["total_sales"].sum().sort_values(ascending=False).head(3)
```

```python
avg_rating =df.groupby("category")["rating"].mean()
print(avg_rating)
```

```python
def rating_label(r):
  if r >=4:
    return"Good"
  elif r ==3:
    return"Average"
  else:
    return"poor"
df["rating_label"]=df["rating"].apply(rating_label)
df.head()
```

Part D Machine Learning

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error
```

```python
x=df[["price" , "quantity" , "order_frequency"]]
y=df["total_sales"]
```

```python
x_train, x_test, y_train , y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
y_pred=model.predict(x_test)
print(y_pred)
```

```python
r2 = r2_score(y_test, y_pred)
print("R2 score:", r2)
```

```python
mae = mean_absolute_error(y_test, y_pred)
print("Mean absolute error:", mae)
```

```python
from sklearn.metrics import mean_squared_error
mse = mean_squared_error(y_test, y_pred)
print("Mean Squared Error:", mse)
```

```python
# Using regression metrics with the correct test variable 'y_test'
print("R2 Score:", r2_score(y_test, y_pred))
print("Mean Absolute Error:", mean_absolute_error(y_test, y_pred))
print("Mean Squared Error:", mean_squared_error(y_test, y_pred))
```
